```python
import pandas as pd
import numpy as np
import glob
import os

# ============================================================
# CIC-IDS2017 PREPROCESSING
# Binary anomaly detection:
# BENIGN = 0
# ATTACK = 1
# ============================================================

files = sorted(glob.glob('/content/*.csv'))

print("CIC-IDS2017 files found:", len(files))

# We will collect a manageable sample from each class.
# This prevents the 2.8M-row dataset from consuming all RAM.
MAX_BENIGN_PER_FILE = 30000
MAX_ATTACK_PER_FILE = 30000

benign_parts = []
attack_parts = []

for file in files:

    print("\nProcessing:", os.path.basename(file))

    benign_count = 0
    attack_count = 0

    for chunk in pd.read_csv(
        file,
        chunksize=50000,
        low_memory=False
    ):

        # Clean column names
        chunk.columns = chunk.columns.str.strip()

        # Make sure Label exists
        if "Label" not in chunk.columns:
            print("WARNING: Label column not found")
            continue

        # Clean labels
        chunk["Label"] = chunk["Label"].astype(str).str.strip()

        # Replace infinite values
        chunk.replace([np.inf, -np.inf], np.nan, inplace=True)

        # Convert binary target
        chunk["target"] = (
            chunk["Label"].str.upper() != "BENIGN"
        ).astype(np.int8)

        # Separate benign and attack traffic
        benign = chunk[chunk["target"] == 0]
        attack = chunk[chunk["target"] == 1]

        # Collect benign samples
        if benign_count < MAX_BENIGN_PER_FILE:
            remaining = MAX_BENIGN_PER_FILE - benign_count
            sample = benign.head(remaining)

            if len(sample) > 0:
                benign_parts.append(sample)
                benign_count += len(sample)

        # Collect attack samples
        if attack_count < MAX_ATTACK_PER_FILE:
            remaining = MAX_ATTACK_PER_FILE - attack_count
            sample = attack.head(remaining)

            if len(sample) > 0:
                attack_parts.append(sample)
                attack_count += len(sample)

        # Stop reading this file when enough samples collected
        if (
            benign_count >= MAX_BENIGN_PER_FILE
            and attack_count >= MAX_ATTACK_PER_FILE
        ):
            break

    print("  Benign collected:", benign_count)
    print("  Attack collected:", attack_count)


# Combine sampled data
print("\nCombining samples...")

benign_df = pd.concat(benign_parts, ignore_index=True)
attack_df = pd.concat(attack_parts, ignore_index=True)

print("Benign total:", len(benign_df))
print("Attack total:", len(attack_df))

# Balance the two classes
min_count = min(len(benign_df), len(attack_df))

benign_df = benign_df.sample(
    n=min_count,
    random_state=42
)

attack_df = attack_df.sample(
    n=min_count,
    random_state=42
)

# Combine
cicids_df = pd.concat(
    [benign_df, attack_df],
    ignore_index=True
)

# Shuffle
cicids_df = cicids_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("\n==============================")
print("FINAL PREPROCESSED DATASET")
print("==============================")
print("Rows:", len(cicids_df))
print("Columns:", len(cicids_df.columns))

print("\nTarget distribution:")
print(cicids_df["target"].value_counts())

print("\nTarget percentages:")
print(cicids_df["target"].value_counts(normalize=True) * 100)

print("\nLabels:")
print(cicids_df["Label"].value_counts().head(20))
```

```python
# ============================================================
# STEP 2 — FEATURE PREPROCESSING
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# Make a copy
df = cicids_df.copy()

print("Original shape:", df.shape)

# ------------------------------------------------------------
# 1. Remove columns that should NOT be used for ML
# ------------------------------------------------------------

columns_to_drop = [
    "Label",
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp"
]

# Drop only columns that actually exist
columns_to_drop = [
    col for col in columns_to_drop
    if col in df.columns
]

X = df.drop(columns=columns_to_drop + ["target"])
y = df["target"]

print("\nDropped columns:")
print(columns_to_drop)

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)


# ------------------------------------------------------------
# 2. Convert everything to numeric
# ------------------------------------------------------------

X = X.apply(pd.to_numeric, errors="coerce")

print("\nNumeric feature shape:", X.shape)


# ------------------------------------------------------------
# 3. Replace infinite values
# ------------------------------------------------------------

X = X.replace([np.inf, -np.inf], np.nan)


# ------------------------------------------------------------
# 4. Remove columns containing only NaN
# ------------------------------------------------------------

all_nan_columns = X.columns[X.isna().all()].tolist()

if all_nan_columns:
    print("\nRemoving all-NaN columns:")
    print(all_nan_columns)

    X = X.drop(columns=all_nan_columns)


# ------------------------------------------------------------
# 5. Fill missing values using median
# ------------------------------------------------------------

imputer = SimpleImputer(strategy="median")

X_imputed = imputer.fit_transform(X)

X = pd.DataFrame(
    X_imputed,
    columns=X.columns
)

print("\nMissing values after imputation:",
      X.isna().sum().sum())


# ------------------------------------------------------------
# 6. Train/Test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n==============================")
print("TRAIN / TEST SPLIT")
print("==============================")

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train distribution:")
print(y_train.value_counts())

print("\ny_test distribution:")
print(y_test.value_counts())
```

```python
# ============================================================
# STEP 3 — CIC-IDS2017 ANOMALY DETECTION MODEL
# Random Forest Classifier
# ============================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("Training CIC-IDS2017 Random Forest model...")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
print("Number of features:", X_train.shape[1])

# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

cicids_model = RandomForestClassifier(
    n_estimators=150,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced",
    max_features="sqrt"
)

# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

cicids_model.fit(X_train, y_train)

print("\nModel training completed successfully.")

# ------------------------------------------------------------
# Predictions
# ------------------------------------------------------------

y_pred = cicids_model.predict(X_test)
y_prob = cicids_model.predict_proba(X_test)[:, 1]

# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("\n==========================================")
print("CIC-IDS2017 ANOMALY DETECTION RESULTS")
print("==========================================")

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

# ------------------------------------------------------------
# Classification report
# ------------------------------------------------------------

print("\n==========================================")
print("CLASSIFICATION REPORT")
print("==========================================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Benign", "Attack"]
    )
)

# ------------------------------------------------------------
# Confusion matrix
# ------------------------------------------------------------

print("\n==========================================")
print("CONFUSION MATRIX")
print("==========================================")

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# ============================================================
# STEP 4 — CIC-IDS2017 MODEL VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, RocCurveDisplay

# ------------------------------------------------------------
# Confusion Matrix
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Benign", "Attack"],
    cmap="Blues"
)

plt.title("CIC-IDS2017 - Confusion Matrix")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# ROC Curve
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

RocCurveDisplay.from_predictions(
    y_test,
    y_prob,
    name="Random Forest"
)

plt.title("CIC-IDS2017 - ROC Curve")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 6 — CICIDS2017 DETECTION REPORT
# ============================================================

import pandas as pd
import numpy as np
import os

print("Creating CICIDS2017 detection report...")

# ------------------------------------------------------------
# 1. Get the original CICIDS labels for the test records
# ------------------------------------------------------------

actual_labels = cicids_df.loc[X_test.index, "Label"].values

# ------------------------------------------------------------
# 2. Convert numeric predictions into readable labels
# ------------------------------------------------------------

predicted_labels = np.where(
    y_pred == 1,
    "ATTACK",
    "BENIGN"
)

# ------------------------------------------------------------
# 3. Attack probability
# ------------------------------------------------------------

attack_probability = y_prob * 100

# ------------------------------------------------------------
# 4. Detection status
# ------------------------------------------------------------

detection_status = np.where(
    y_pred == 1,
    "Detected",
    "Normal"
)

# ------------------------------------------------------------
# 5. Severity
# ------------------------------------------------------------

def calculate_severity(probability):
    if probability >= 80:
        return "High"
    elif probability >= 50:
        return "Medium"
    else:
        return "Low"

severity = [
    calculate_severity(prob)
    for prob in attack_probability
]

# ------------------------------------------------------------
# 6. Create final report
# ------------------------------------------------------------

cicids_report = pd.DataFrame({
    "Actual": actual_labels,
    "Predicted": predicted_labels,
    "Attack_Probability": np.round(attack_probability, 2),
    "Detection_Status": detection_status,
    "Severity": severity
})

# ------------------------------------------------------------
# 7. Display report information
# ------------------------------------------------------------

print("\n==============================================")
print("CICIDS2017 ANOMALY DETECTION REPORT")
print("==============================================")

print("\nReport shape:")
print(cicids_report.shape)

print("\nReport columns:")
print(cicids_report.columns.tolist())

print("\nFirst 10 records:")
display(cicids_report.head(10))

# ------------------------------------------------------------
# 8. Severity distribution
# ------------------------------------------------------------

print("\nSeverity counts:")
print(cicids_report["Severity"].value_counts())

# ------------------------------------------------------------
# 9. Detection status distribution
# ------------------------------------------------------------

print("\nDetection Status counts:")
print(cicids_report["Detection_Status"].value_counts())

# ------------------------------------------------------------
# 10. Actual label distribution
# ------------------------------------------------------------

print("\nActual CICIDS labels:")
print(cicids_report["Actual"].value_counts())
```

```python
# ============================================================
# STEP 4 — CICIDS2017 SUMMARY REPORT
# ============================================================

import pandas as pd
import numpy as np

# Use the verified saved report
df_report = verified_report.copy()

# ------------------------------------------------------------
# 1. Basic counts
# ------------------------------------------------------------

total_records = len(df_report)

# Actual labels
actual_normal = (df_report["Actual"] == "BENIGN").sum()
actual_attack = (df_report["Actual"] != "BENIGN").sum()

# Predicted labels
predicted_normal = (df_report["Predicted"] == "BENIGN").sum()
predicted_attack = (df_report["Predicted"] == "ATTACK").sum()

# ------------------------------------------------------------
# 2. Correct / incorrect predictions
# ------------------------------------------------------------

# Actual binary class
actual_binary = np.where(
    df_report["Actual"] == "BENIGN",
    "BENIGN",
    "ATTACK"
)

predicted_binary = df_report["Predicted"].values

# True negatives
true_negative = np.sum(
    (actual_binary == "BENIGN") &
    (predicted_binary == "BENIGN")
)

# True positives
true_positive = np.sum(
    (actual_binary == "ATTACK") &
    (predicted_binary == "ATTACK")
)

# False positives
false_positive = np.sum(
    (actual_binary == "BENIGN") &
    (predicted_binary == "ATTACK")
)

# False negatives
false_negative = np.sum(
    (actual_binary == "ATTACK") &
    (predicted_binary == "BENIGN")
)

# ------------------------------------------------------------
# 3. Performance metrics
# ------------------------------------------------------------

accuracy = (
    (true_positive + true_negative) / total_records
    if total_records > 0 else 0
)

precision = (
    true_positive / (true_positive + false_positive)
    if (true_positive + false_positive) > 0 else 0
)

recall = (
    true_positive / (true_positive + false_negative)
    if (true_positive + false_negative) > 0 else 0
)

f1_score = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) > 0 else 0
)

# ------------------------------------------------------------
# 4. Print summary
# ------------------------------------------------------------

print("============================================================")
print("CICIDS2017 AI ANOMALY DETECTION SUMMARY")
print("============================================================")

print("\nDataset Summary")
print("----------------------------")
print("Total records       :", total_records)
print("Actual Normal       :", actual_normal)
print("Actual Attack       :", actual_attack)

print("\nPrediction Summary")
print("----------------------------")
print("Predicted Normal    :", predicted_normal)
print("Predicted Attack    :", predicted_attack)

print("\nConfusion Matrix Values")
print("----------------------------")
print("True Negatives      :", true_negative)
print("True Positives      :", true_positive)
print("False Positives     :", false_positive)
print("False Negatives     :", false_negative)

print("\nPerformance Metrics")
print("----------------------------")
print("Accuracy            :", round(accuracy * 100, 2), "%")
print("Precision           :", round(precision * 100, 2), "%")
print("Recall              :", round(recall * 100, 2), "%")
print("F1-Score            :", round(f1_score * 100, 2), "%")

print("\nSeverity Distribution")
print("----------------------------")
print(df_report["Severity"].value_counts())

print("\nAttack-Type Distribution")
print("----------------------------")
print(
    df_report.loc[
        df_report["Actual"] != "BENIGN",
        "Actual"
    ].value_counts()
)

print("\n============================================================")
print("SUMMARY COMPLETE")
print("============================================================")
```

```python
# ============================================================
# CICIDS2017 — FINAL DYNAMIC EVALUATION
# ============================================================

import os
import json
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve
)

# ------------------------------------------------------------
# Results folder
# ------------------------------------------------------------

results_path = (
    "/content/drive/MyDrive/"
    "NetShield_AI_Milestone_2/results/CICIDS2017"
)

os.makedirs(results_path, exist_ok=True)


# ------------------------------------------------------------
# 1. Use only records with valid Actual labels
# ------------------------------------------------------------

evaluation_df = df_report[
    df_report["Actual"].notna()
].copy()

print("============================================================")
print("CICIDS2017 FINAL MODEL EVALUATION")
print("============================================================")

print("\nTotal records        :", len(df_report))
print("Valid labeled records:", len(evaluation_df))
print(
    "Missing-label records:",
    len(df_report) - len(evaluation_df)
)


# ------------------------------------------------------------
# 2. Convert Actual and Predicted to binary labels
# ------------------------------------------------------------

y_true_final = np.where(
    evaluation_df["Actual"].astype(str).str.strip() == "BENIGN",
    0,
    1
)

y_pred_final = np.where(
    evaluation_df["Predicted"].astype(str).str.strip() == "BENIGN",
    0,
    1
)


# ------------------------------------------------------------
# 3. Get attack probabilities
# ------------------------------------------------------------

y_prob_final = (
    evaluation_df["Attack_Probability"]
    .astype(float)
    .values
    / 100.0
)


# ------------------------------------------------------------
# 4. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_true_final,
    y_pred_final,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()

print("\nConfusion Matrix")
print("----------------------------")
print("True Negatives :", tn)
print("True Positives :", tp)
print("False Positives:", fp)
print("False Negatives:", fn)


# ------------------------------------------------------------
# 5. Performance Metrics
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true_final,
    y_pred_final
)

precision = precision_score(
    y_true_final,
    y_pred_final,
    zero_division=0
)

recall = recall_score(
    y_true_final,
    y_pred_final,
    zero_division=0
)

f1 = f1_score(
    y_true_final,
    y_pred_final,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true_final,
    y_prob_final
)

print("\nPerformance Metrics")
print("----------------------------")
print(f"Accuracy : {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall   : {recall * 100:.2f}%")
print(f"F1-Score : {f1 * 100:.2f}%")
print(f"ROC-AUC  : {roc_auc:.4f}")


# ------------------------------------------------------------
# 6. Save Confusion Matrix
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.imshow(cm)

plt.title("CIC-IDS2017 - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")

plt.xticks(
    [0, 1],
    ["Benign", "Attack"]
)

plt.yticks(
    [0, 1],
    ["Benign", "Attack"]
)

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()
plt.tight_layout()

cm_path = os.path.join(
    results_path,
    "confusion_matrix.png"
)

plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\n✅ Confusion matrix updated!")
print(cm_path)


# ------------------------------------------------------------
# 7. Save ROC Curve
# ------------------------------------------------------------

fpr, tpr, thresholds = roc_curve(
    y_true_final,
    y_prob_final
)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Random Forest (AUC = {roc_auc:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.title("CIC-IDS2017 - ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend(loc="lower right")
plt.grid(True)
plt.tight_layout()

roc_path = os.path.join(
    results_path,
    "roc_curve.png"
)

plt.savefig(
    roc_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\n✅ ROC curve updated!")
print(roc_path)


# ------------------------------------------------------------
# 8. Save Metrics JSON
# ------------------------------------------------------------

cic_metrics = {
    "Dataset": "CIC-IDS2017",
    "Model": "Random Forest",

    "Total Records": int(len(df_report)),
    "Valid Labeled Records": int(len(evaluation_df)),
    "Missing Label Records": int(
        len(df_report) - len(evaluation_df)
    ),

    "True Negatives": int(tn),
    "True Positives": int(tp),
    "False Positives": int(fp),
    "False Negatives": int(fn),

    "Accuracy": round(float(accuracy), 4),
    "Precision": round(float(precision), 4),
    "Recall": round(float(recall), 4),
    "F1-Score": round(float(f1), 4),
    "ROC-AUC": round(float(roc_auc), 4)
}

metrics_file = os.path.join(
    results_path,
    "metrics.json"
)

with open(metrics_file, "w") as f:
    json.dump(
        cic_metrics,
        f,
        indent=4
    )

print("\n✅ Metrics JSON updated!")
print(metrics_file)


# ------------------------------------------------------------
# 9. Final Verification
# ------------------------------------------------------------

print("\n============================================================")
print("CICIDS2017 RESULTS UPDATED SUCCESSFULLY")
print("============================================================")

print("\nFiles updated:")
print("✅ confusion_matrix.png")
print("✅ metrics.json")
print("✅ roc_curve.png")

print("\nFinal values:")
print("Total records        :", len(df_report))
print("Valid labeled records:", len(evaluation_df))
print("Missing labels       :", len(df_report) - len(evaluation_df))
print("True Negatives       :", tn)
print("True Positives       :", tp)
print("False Positives      :", fp)
print("False Negatives      :", fn)
print(f"Accuracy             : {accuracy * 100:.2f}%")
print(f"Precision            : {precision * 100:.2f}%")
print(f"Recall               : {recall * 100:.2f}%")
print(f"F1-Score             : {f1 * 100:.2f}%")
print(f"ROC-AUC              : {roc_auc:.4f}")
```

```python
# ============================================================
# CICIDS2017 — THREAT CLASSIFICATION DATASET
# Dynamic multiclass attack-label preparation
# ============================================================

import pandas as pd
import numpy as np

print("============================================================")
print("CICIDS2017 THREAT CLASSIFICATION DATA PREPARATION")
print("============================================================")

# Use the already prepared CICIDS dataset
threat_df = cicids_df.copy()

# ------------------------------------------------------------
# 1. Clean original labels
# ------------------------------------------------------------

threat_df["Label"] = (
    threat_df["Label"]
    .astype(str)
    .str.strip()
)

# Remove invalid/missing labels
threat_df = threat_df[
    threat_df["Label"].notna() &
    (threat_df["Label"].str.lower() != "nan")
].copy()

# ------------------------------------------------------------
# 2. Display all labels dynamically
# ------------------------------------------------------------

label_counts = threat_df["Label"].value_counts()

print("\nAll CICIDS labels detected:")
print(label_counts)

print("\nNumber of classes detected:", len(label_counts))

# ------------------------------------------------------------
# 3. Check classes with too few records
# ------------------------------------------------------------

minimum_samples_required = 2

rare_labels = label_counts[
    label_counts < minimum_samples_required
]

if len(rare_labels) > 0:

    print("\nWARNING: Classes with fewer than",
          minimum_samples_required,
          "records:")

    print(rare_labels)

    # These classes cannot safely be stratified into
    # train/test sets.
    threat_df = threat_df[
        ~threat_df["Label"].isin(rare_labels.index)
    ].copy()

else:

    print("\nNo classes need to be removed.")

# ------------------------------------------------------------
# 4. Final label distribution
# ------------------------------------------------------------

final_label_counts = threat_df["Label"].value_counts()

print("\n============================================================")
print("FINAL THREAT CLASS DISTRIBUTION")
print("============================================================")

print(final_label_counts)

print("\nTotal records:", len(threat_df))
print("Total classes:", len(final_label_counts))

print("\nClass names:")
for label in sorted(final_label_counts.index):
    print(" -", label)
```

```python
# ============================================================
# CICIDS2017 — THREAT CLASSIFICATION FEATURES
# Dynamic feature preparation
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

print("============================================================")
print("THREAT CLASSIFICATION FEATURE PREPARATION")
print("============================================================")

# ------------------------------------------------------------
# 1. Separate target
# ------------------------------------------------------------

y_threat = threat_df["Label"].copy()

# ------------------------------------------------------------
# 2. Remove non-ML columns dynamically
# ------------------------------------------------------------

excluded_columns = [
    "Label",
    "target",
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp"
]

feature_columns = [
    col
    for col in threat_df.columns
    if col not in excluded_columns
]

print("\nCandidate feature columns:", len(feature_columns))

# ------------------------------------------------------------
# 3. Create feature dataframe
# ------------------------------------------------------------

X_threat_raw = threat_df[feature_columns].copy()

# Convert features to numeric
X_threat_raw = X_threat_raw.apply(
    pd.to_numeric,
    errors="coerce"
)

# Replace infinite values
X_threat_raw = X_threat_raw.replace(
    [np.inf, -np.inf],
    np.nan
)

# ------------------------------------------------------------
# 4. Remove columns containing only NaN
# ------------------------------------------------------------

all_nan_columns = X_threat_raw.columns[
    X_threat_raw.isna().all()
].tolist()

if all_nan_columns:

    print("\nRemoving all-NaN columns:")

    for col in all_nan_columns:
        print(" -", col)

    X_threat_raw = X_threat_raw.drop(
        columns=all_nan_columns
    )

# ------------------------------------------------------------
# 5. Remove duplicate feature columns if any
# ------------------------------------------------------------

duplicate_columns = X_threat_raw.columns[
    X_threat_raw.columns.duplicated()
].tolist()

if duplicate_columns:

    print("\nRemoving duplicate feature columns:")
    print(duplicate_columns)

    X_threat_raw = X_threat_raw.loc[
        :,
        ~X_threat_raw.columns.duplicated()
    ]

print("\nFinal feature count:",
      X_threat_raw.shape[1])

# ------------------------------------------------------------
# 6. Train/test split BEFORE fitting imputer
# ------------------------------------------------------------

X_train_threat, X_test_threat, y_train_threat, y_test_threat = (
    train_test_split(
        X_threat_raw,
        y_threat,
        test_size=0.20,
        random_state=42,
        stratify=y_threat
    )
)

print("\nTraining records:", len(X_train_threat))
print("Testing records :", len(X_test_threat))

print("\nTraining classes:",
      y_train_threat.nunique())

print("Testing classes :",
      y_test_threat.nunique())

# ------------------------------------------------------------
# 7. Fit imputer ONLY on training data
# ------------------------------------------------------------

threat_imputer = SimpleImputer(
    strategy="median"
)

X_train_threat_imputed = threat_imputer.fit_transform(
    X_train_threat
)

X_test_threat_imputed = threat_imputer.transform(
    X_test_threat
)

# Convert back to DataFrame
threat_feature_names = list(
    X_train_threat.columns
)

X_train_threat = pd.DataFrame(
    X_train_threat_imputed,
    columns=threat_feature_names,
    index=X_train_threat.index
)

X_test_threat = pd.DataFrame(
    X_test_threat_imputed,
    columns=threat_feature_names,
    index=X_test_threat.index
)

print("\nFinal training shape:",
      X_train_threat.shape)

print("Final testing shape :",
      X_test_threat.shape)

print("\nMissing values in training:",
      X_train_threat.isna().sum().sum())

print("Missing values in testing:",
      X_test_threat.isna().sum().sum())
```

```python
# ============================================================
# CICIDS2017 — RISK SCORING SYSTEM
# STEP 7
# ============================================================

import numpy as np
import pandas as pd

print("=" * 60)
print("CICIDS2017 RISK SCORING SYSTEM")
print("=" * 60)

# ------------------------------------------------------------
# 1. Use the existing threat classification report
# ------------------------------------------------------------

risk_report = cicids_threat_report.copy()

# ------------------------------------------------------------
# 2. Validate required columns
# ------------------------------------------------------------

required_columns = [
    "Actual_Threat",
    "Predicted_Threat",
    "Threat_Probability",
    "Threat_Level"
]

missing_columns = [
    column
    for column in required_columns
    if column not in risk_report.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("\nRequired threat-classification columns verified:")
for column in required_columns:
    print("✅", column)

# ------------------------------------------------------------
# 3. Dynamic risk-score calculation
#
# BENIGN:
#     Low base risk because it is predicted as normal traffic.
#
# ATTACK:
#     Risk is based on the model's confidence.
#
# Score range:
#     0 - 100
# ------------------------------------------------------------

def calculate_risk_score(row):

    predicted_threat = str(
        row["Predicted_Threat"]
    ).strip()

    probability = float(
        row["Threat_Probability"]
    )

    # --------------------------------------------------------
    # BENIGN traffic
    # --------------------------------------------------------

    if predicted_threat.upper() == "BENIGN":

        score = probability * 0.10

    # --------------------------------------------------------
    # Any detected attack
    # --------------------------------------------------------

    else:

        score = probability

    # Keep score safely within 0–100
    score = max(
        0,
        min(100, score)
    )

    return round(score, 2)


risk_report["Risk_Score"] = (
    risk_report.apply(
        calculate_risk_score,
        axis=1
    )
)

# ------------------------------------------------------------
# 4. Dynamic risk-level classification
# ------------------------------------------------------------

def calculate_risk_level(score):

    if score >= 80:
        return "Critical"

    elif score >= 60:
        return "High"

    elif score >= 30:
        return "Medium"

    else:
        return "Low"


risk_report["Risk_Level"] = (
    risk_report["Risk_Score"].apply(
        calculate_risk_level
    )
)

# ------------------------------------------------------------
# 5. Display final report structure
# ------------------------------------------------------------

print("\n============================================================")
print("CICIDS2017 RISK SCORING REPORT")
print("============================================================")

print("\nReport shape:")
print(risk_report.shape)

print("\nReport columns:")
print(risk_report.columns.tolist())

# ------------------------------------------------------------
# 6. First 10 records
# ------------------------------------------------------------

print("\n============================================================")
print("FIRST 10 RISK RECORDS")
print("============================================================")

display(
    risk_report.head(10)
)

# ------------------------------------------------------------
# 7. Risk score statistics
# ------------------------------------------------------------

print("\n============================================================")
print("RISK SCORE STATISTICS")
print("============================================================")

print(
    risk_report["Risk_Score"].describe()
)

# ------------------------------------------------------------
# 8. Risk-level distribution
# ------------------------------------------------------------

print("\n============================================================")
print("RISK LEVEL DISTRIBUTION")
print("============================================================")

print(
    risk_report["Risk_Level"].value_counts()
)

# ------------------------------------------------------------
# 9. Predicted threat distribution
# ------------------------------------------------------------

print("\n============================================================")
print("THREAT-WISE RISK DISTRIBUTION")
print("============================================================")

threat_risk_summary = (
    risk_report
    .groupby("Predicted_Threat")["Risk_Score"]
    .agg(
        Count="count",
        Average_Risk_Score="mean",
        Maximum_Risk_Score="max"
    )
    .sort_values(
        "Average_Risk_Score",
        ascending=False
    )
)

threat_risk_summary[
    "Average_Risk_Score"
] = (
    threat_risk_summary[
        "Average_Risk_Score"
    ].round(2)
)

threat_risk_summary[
    "Maximum_Risk_Score"
] = (
    threat_risk_summary[
        "Maximum_Risk_Score"
    ].round(2)
)

print(
    threat_risk_summary
)

# ------------------------------------------------------------
# 10. Attack-only risk analysis
# ------------------------------------------------------------

attack_risk_report = risk_report[
    risk_report["Predicted_Threat"].str.upper() != "BENIGN"
].copy()

print("\n============================================================")
print("ATTACK RISK ANALYSIS")
print("============================================================")

print(
    "Total predicted attack records:",
    len(attack_risk_report)
)

if len(attack_risk_report) > 0:

    print(
        "Average attack risk score:",
        round(
            attack_risk_report[
                "Risk_Score"
            ].mean(),
            2
        )
    )

    print(
        "Maximum attack risk score:",
        round(
            attack_risk_report[
                "Risk_Score"
            ].max(),
            2
        )
    )

# ------------------------------------------------------------
# 11. Verification
# ------------------------------------------------------------

print("\n============================================================")
print("RISK SCORE VERIFICATION")
print("============================================================")

score_values_valid = (
    risk_report["Risk_Score"]
    .between(0, 100)
    .all()
)

risk_levels_valid = (
    risk_report["Risk_Level"]
    .isin(
        [
            "Low",
            "Medium",
            "High",
            "Critical"
        ]
    )
    .all()
)

no_missing_scores = (
    risk_report["Risk_Score"]
    .notna()
    .all()
)

no_missing_levels = (
    risk_report["Risk_Level"]
    .notna()
    .all()
)

record_count_matches = (
    len(risk_report)
    == len(y_test_threat)
    == len(threat_predictions)
    == len(threat_probabilities)
)

print(
    "All risk scores between 0 and 100:",
    score_values_valid
)

print(
    "All risk levels valid:",
    risk_levels_valid
)

print(
    "No missing risk scores:",
    no_missing_scores
)

print(
    "No missing risk levels:",
    no_missing_levels
)

print(
    "Risk records:",
    len(risk_report)
)

print(
    "Prediction/report record counts match:",
    record_count_matches
)

# ------------------------------------------------------------
# 12. Final verification
# ------------------------------------------------------------

all_risk_checks_passed = (
    score_values_valid
    and risk_levels_valid
    and no_missing_scores
    and no_missing_levels
    and record_count_matches
)

print("\nAll risk-scoring checks passed:")
print(all_risk_checks_passed)

print("\n============================================================")

if all_risk_checks_passed:

    print(
        "✅ CICIDS2017 RISK SCORING COMPLETE"
    )

else:

    print(
        "❌ CICIDS2017 RISK SCORING NEEDS ATTENTION"
    )

print("============================================================")
```
